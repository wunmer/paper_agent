3. 决策曲线分析为什么使用净获益评价模型？

回答： ## 结论

决策曲线分析（DCA）选择净获益作为评价尺度，核心原因是：**其他常用指标只衡量“预测得准不准”，不衡量“按这个预测做决策的后果好不好”，而净获益直接把临床后果（真阳性、假阳性的相对代价）纳入同一个可比较的量纲。**

## 具体理由（据 Vickers & Elkin 论文）

**1. 准确度类指标不含后果信息，无法回答“模型值不值得用”**

AUC 被解释为“随机一对人中，发生事件者预测概率更高”的概率，是单一的比较统计量；但它“只关注模型的预测准确性，因此无法告诉我们模型是否值得使用，或两个模型中哪个更可取”（p2）。论文明确指出：敏感度、特异度、AUC 这类准确度指标“并不反映模型的临床价值”，**AUC 更高的模型有时导致更差的结局，而 AUC 相差很大的模型临床上可以相当**（p8）。文中实证例子里，Gleason 规则 AUC 仅 0.56，却在关键的 30%–60% 阈值区间与 AUC 0.73 的多变量模型相当，并优于 AUC 0.58 的 stage 规则（p7）——这类现象只能靠净获益揭示。

**2. 阈值概率刻画了假阳性与假阴性的相对危害，净获益是这一权衡的量化**

DCA 的出发假设是：患者/医生愿意接受治疗的阈值概率 p_t 本身携带了“他如何权衡假阳性与假阴性危害”的信息——由决策树推导可得，p_t 反映假阴性相对假阳性的危害比；因此 p_t = 10% 意味着该医生认为漏诊一个阳性其危害约为不必要治疗一个阴性的 9 倍（p3, p4）。净获益公式（Peirce 公式）为：

净获益 = 真阳性数/n − 假阳性数/n × [p_t/(1−p_t)]

即“从真阳性比例中扣除假阳性比例，并按两者的相对危害加权”（p4）。

**3. 净获益具有临床可解释性，是“可换算成事件数”的单位**

例如在 5% 阈值下净获益 0.062，可解释为：与“假定所有患者阴性”相比，用该模型“相当于每 100 名患者净多获得 6.2 个真阳性，且不增加假阳性”；两模型净获益差 0.013 则等价于每 100 人中减少 25 例假阳性（p6）。这种解释方式“与临床试验的解读可比”：获益明确就用，明确无效就不用，只对部分人有效则应个体化讨论（p6）。

**4. 净获益提供了明确的对照基准，使“好坏”可判定**

净获益的理论范围从负无穷到疾病发生率（p4）。因为“全部阴性/不治疗”策略的净获益为 0，模型净获益为正即优于不治疗；而“全部治疗”策略可用同一公式算出（例中 p_t=10% 时为 −0.0039，低于模型的 0.0443），故模型可同时与两个默认策略比较（p4）。正是这种“有参照点的差值”，让 DCA 能给出模型有价值的阈值区间、获益幅度，以及多个模型中哪个最优（p1, p6）。

**5. 相比完整决策分析，净获益所需输入更少、更易落地**

完整决策分析需要治疗的成本与效果、患者对健康状态的效用等外部信息；净获益方法则“不需要治疗成本或效果信息、也不需要患者对不同健康状态的赋值”，可直接用于模型验证数据集（p8）。只有当检查本身有创或昂贵时才需额外设定 “test harm”（例如漏诊危害是接受检查危害的 50 倍则记为 0.02）（p6）。

**6. 阈值依赖假设的边界**

净获益方法把 p_t 当作因人而异的量，因此不做“单一最优阈值”的假设：只有当治疗获益危害清晰可估、且个体间对获益危害的评价相近时，单一阈值才成立；否则更宜给出整条决策曲线（p8）。此外论文也指出未处理预测不确定性，置信带等方法当时仍在研究中（p9）。

4. 构建预测模型时，确定样本量需要考虑哪些因素？

5轮调用

回答： ## 结论：不应再用“每个变量10个事件”的简单规则；应综合考虑参与者数 n、事件数 E 与预测参数数 p

### 需要一并考虑的因素

1. **候选预测变量的个数，以及它们对应的“预测参数个数 p”**
   一个分类变量（比如有3个及以上类别）或按非线性趋势建模的连续变量会占用多个参数；交互项也要计入。因此要用**每个预测参数的事件数（EPP）**，而不是每个变量的事件数 [minimum sample size | PDF第2页 | minimum sample size-p2-c2]。

2. **预期模型表现（Cox-Snell R²），并由此决定模型能解释的变异大小**
   事先设定合理的 R²_adj（可借鉴同一领域已发表模型的 C 统计量或 R²），再据此设定 max(R²_app)，它由结局的总发生率或总速率决定 [minimum sample size | PDF第12页 | minimum sample size-p12-c1]。R² 同时反映了结局比例与模型整体拟合程度，因此样本量计算能贴合具体模型与场景，而不是套用统一的 EPP 规则 [minimum sample size | PDF第17页 | minimum sample size-p17-c4]。

3. **可接受的过拟合程度（全局收缩因子 S）**
   准则(i)：要求收缩因子 ≥ 0.90（预期过拟合约10%），由它决定最小 n 与 E [minimum sample size | PDF第12页 | minimum sample size-p12-c1][minimum sample size | PDF第2页 | minimum sample size-p2-c4]。收缩因子从0.90提高到0.95时，样本量会增加很多；S<0.9 会带来更大的过拟合，因此不推荐 [minimum sample size | PDF第16页 | minimum sample size-p16-c1]。

4. **可接受的乐观程度（表观与校正 Nagelkerke R² 的绝对差 < 0.05）**
   准则(ii)：先反算出所需的收缩因子，再据此重算样本量；如果它比准则(i)所用的 S 更严格，就需要更大的样本量 [minimum sample size | PDF第12页 | minimum sample size-p12-c2]。

5. **总体风险（或模型截距）的估计精度**
   准则(iii)：整体风险要达到绝对误差 < 0.05；二分类结局用相应公式，生存结局用指数模型近似 [minimum sample size | PDF第12页 | minimum sample size-p12-c2][minimum sample size | PDF第11页 | minimum sample size-p11-c1]。

6. **最终样本量取三条准则算出的最大值**
   Step 6：最小样本量取步骤3–5结果中的最大者，这样三条准则才都能满足 [minimum sample size | PDF第12页 | minimum sample size-p12-c2]。

7. **生存结局（time-to-event）还需考虑随访时长、删失率和事件率**
   在同样的样本量下，所需人年数 = n × 平均随访年数；事件数 = 人年数 × 事件率。因此这些假设会直接改变可行性 [minimum sample size | PDF第15页 | minimum sample size-p15-c2][minimum sample size | PDF第15页 | minimum sample size-p15-c3]。

8. **数据本身的复杂结构**
   预测效应的大小、结局的总风险、预测变量的分布、各分类水平内的事件数都会影响所需样本量；此外还要注意“分离”（某个预测变量或预测变量的线性组合能完全分开事件与非事件）问题，这可能需要更大的 EPP [minimum sample size | PDF第2页 | minimum sample size-p2-c3][minimum sample size | PDF第18页 | minimum sample size-p18-c4]。

### 当样本量无法达到要求时

建议的做法是**减少候选预测参数 p**（例如依据既往系统评价筛选变量），或采用更保守（更大）的 R²_adj 假设；不建议把 S 降到0.9以下，因为那意味着更大的过拟合 [minimum sample size | PDF第15页 | minimum sample size-p15-c3]。

### 报告要求

TRIPOD 第8条要求作者说明研究样本量是如何确定的 [transparent multivariable prediction | PDF第6页 | transparent multivariable prediction-p6-c3]；局限性讨论中也应汇报“每个预测参数的事件数少”等问题 [transparent multivariable prediction | PDF第7页 | transparent multivariable prediction-p7-c1]。

5. TRIPOD与TRIPOD+AI在报告要求上有哪些区别？

6轮模型调用

回答： ## 总体关系

TRIPOD+AI（2024）是对 TRIPOD 2015 的**更新版**，并非补充版：文中明确写明新清单**取代（supersedes）TRIPOD 2015 清单，后者不应再使用**，只有大部分未变或仅微调的项目被保留 [AI for upgrated guidance | p8 | AI for upgrated guidance-p8-c1]。

## 主要体现在以下几方面的区别

**1. 适用范围与建模方法**
- TRIPOD 2015：主要针对以回归建模开发的模型（“largely agnostic to the type of modelling approach”），但未专门覆盖机器学习 [AI for upgrated guidance | p2 | AI for upgrated guidance-p2-c3、p2-c4]。
- TRIPOD+AI：覆盖使用**任何回归或机器学习方法**（随机森林、深度学习等）的预测模型研究；并指出机器学习模型往往没有简单方程、所用预测变量甚至不明确，因此需要有 TRIPOD 2015 未涵盖的额外报告内容 [AI for upgrated guidance | p2 | AI for upgrated guidance-p2-c4、p2-c6]。
- 同时统一（harmonise）回归与机器学习两个领域的术语体系 [AI for upgrated guidance | p8 | AI for upgrated guidance-p8-c1]。

**2. 清单结构与规模**
- TRIPOD 2015：22 个条目——标题与摘要（1–2）、背景与目标（3）、方法（4–12）、结果（13–17）、讨论（18–20）、其他信息（21–22）；条目按 D（开发）/ V（验证）/ D;V 标注 [transparent multivariable prediction | p3 | transparent multivariable prediction-p3-c7；p6-c1]（该文 p3 亦说明哪些条目仅适用于开发、哪些仅适用于（外部）验证）[transparent multivariable prediction | p3 | transparent multivariable prediction-p3-c1]。
- TRIPOD+AI：**27 个主条目、共 52 个子条目**；分节为标题（1）、摘要（2）、前言（3–4）、方法（5–17）、**开放科学（18）**、**患者与公众参与（19）**、结果（20–24）、讨论（25–27）；标注改为 D/E（E 即 evaluation，取代“validation”一词）[AI for upgrated guidance | p7 | AI for upgrated guidance-p7-c6；p3 | AI for upgrated guidance-p3-c3]。
- 注：TRIPOD+AI 文中把 TRIPOD 2015 描述为“37 个条目（25 个通用、6 个开发专用、6 个验证专用）”，与 TRIPOD 原文“22 个条目”口径不同（主条目 vs 子条目计数），引用时需注意 [AI for upgrated guidance | p2 | AI for upgrated guidance-p2-c4]。

**3. Box 2 列出的主要新增/修改内容** [AI for upgrated guidance | p8 | AI for upgrated guidance-p8-c1、p8-c2]
- **公平性（fairness）**贯穿整个清单，要求报告是否采用针对公平性的具体方法；
- **模型性能条目修改**：建议在关键亚组（如社会人口学亚组）中评价模型表现（item 23a）[AI for upgrated guidance | p7 | AI for upgrated guidance-p7-c4]；
- **新增患者与公众参与条目**（item 19）；
- **新增“开放科学”板块**，含研究方案可及性（18c）、注册（18d）、数据共享（18e）、代码共享（18f），另有资助（18a）、利益冲突（18b）[AI for upgrated guidance | p7 | AI for upgrated guidance-p7-c2；p10 | AI for upgrated guidance-p10-c1]；

- 清单改为项目符号结构，并配套扩展版清单（supplementary table 1）以便逐条提供简明说明。

**4. 讨论部分内容**
